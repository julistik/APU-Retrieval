```python
import json
import pandas as pd

with open("test_data.json", "r", encoding="utf-8") as f:
    true_data = json.load(f)

with open("completed_test_data.json", "r", encoding="utf-8") as f:
    pred_data = json.load(f)

print(f"Эталонных примеров: {len(true_data)}")
print(f"Предсказанных примеров: {len(pred_data)}")
assert len(true_data) == len(pred_data), "Количество строк не совпадает!"
```

```python
rows = []

for i, (true_item, pred_item) in enumerate(zip(true_data, pred_data)):
    true_index = true_item.get("index")
    pred_index = pred_item.get("index")

    true_sense = true_item.get("sense_number")
    pred_sense = pred_item.get("sense_number")

    true_meaning = str(true_item.get("meaning", "")).strip().lower()
    pred_meaning = str(pred_item.get("meaning", "")).strip().lower()

    rows.append({
        "row": i,
        "example": true_item.get("example"),

        "true_index": true_index,
        "pred_index": pred_index,
        "index_correct": true_index == pred_index,

        "true_sense": true_sense,
        "pred_sense": pred_sense,
        "sense_correct": true_sense == pred_sense,

        "true_meaning": true_meaning,
        "pred_meaning": pred_meaning,
        "meaning_correct": true_meaning == pred_meaning,

        "all_correct": (
            true_index == pred_index
            and true_sense == pred_sense
            and true_meaning == pred_meaning
        )
    })

df = pd.DataFrame(rows)

errors = df[
    df["sense_correct"] == False
]

errors.to_csv("gpt_errors.csv", index=False, encoding="utf-8-sig")

print("МЕТРИКИ")
print(f"Всего строк: {len(df)}")
print(f"Index accuracy:   {df['index_correct'].mean():.4f}")
print(f"Sense accuracy:   {df['sense_correct'].mean():.4f}")
print(f"All accuracy:     {df['all_correct'].mean():.4f}")
```

```python
print("\nОШИБКИ")
errors_df = df[df["all_correct"] == False]

print(f"Всего ошибочных строк: {len(errors_df)}")

errors_df[50:60]
```
