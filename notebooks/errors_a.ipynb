```python
import pandas as pd

bert = pd.read_csv("bert_errors.csv")
gpt = pd.read_csv("gpt_errors.csv")

bert_examples = set(bert["example"])
gpt_examples = set(gpt["example"])

# ошибки только BERT
only_bert = bert[bert["example"].isin(bert_examples - gpt_examples)]

# ошибки только GPT
only_gpt = gpt[gpt["example"].isin(gpt_examples - bert_examples)]

# ошибки у обоих
both = bert[bert["example"].isin(bert_examples & gpt_examples)]

only_bert.to_csv("only_bert_errors.csv", index=False)
only_gpt.to_csv("only_gpt_errors.csv", index=False)
both.to_csv("both_errors.csv", index=False)

print("Только BERT:", len(only_bert))
print("Только GPT:", len(only_gpt))
print("Оба ошиблись:", len(both))
```
